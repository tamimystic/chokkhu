# Chokkhu: Sovereign End-to-End Tabular Pipeline
A zero-heavy-dependency machine learning library built purely on **NumPy** and Python Standard Library.
- **Zero SciPy, Zero Scikit-learn, Zero PyTorch/TensorFlow**
- **1-Line Sovereign AutoML Pipeline**
- **Full Dynamic Hyperparameter Customization**
- **Pure NumPy Model Evaluation & Feature Importance**

```python
import sys
import numpy as np
import pandas as pd
import chokkhu as ck
print(f'Chokkhu version: {ck.__version__}')
```

## 1. Synthetic Dataset with Missing Values & Categoricals

```python
np.random.seed(42)
n_samples = 200
df = pd.DataFrame({
    'age': np.random.randint(18, 70, size=n_samples).astype(float),
    'salary': np.random.uniform(25000, 120000, size=n_samples),
    'credit_score': np.random.normal(650, 80, size=n_samples),
    'department': np.random.choice(['Tech', 'Sales', 'HR', 'Finance'], size=n_samples),
    'approved': np.random.choice([0, 1], size=n_samples, p=[0.4, 0.6])
})
# Introduce missing values
df.loc[::12, 'age'] = np.nan
df.loc[::15, 'credit_score'] = np.nan
print(f'Dataset shape: {df.shape}')
df.head()
```

## 2. Single-Line Sovereign AutoML
`ck.auto_train` cleans missing values, encodes categoricals, standardizes features, evaluates candidate models, and selects the best model automatically.

```python
automl_result = ck.auto_train(
    df,
    target='approved',
    task='classification',
    time_budget=10
)
print(f'Best Candidate Model: {automl_result.best_model_name}')
print(f'Validation Accuracy: {automl_result.best_score:.4f}')
print('Summary Table:')
display(automl_result.summary())
```

## 3. Dynamic Hyperparameter Customization
Chokkhu provides full dynamic parameter flexibility across all model families.

```python
X = df.drop(columns=['approved', 'department']).fillna(0).values
y = df['approved'].values

# Train an ensemble GradientBoostingClassifier with custom parameters
gb_model = ck.models.ml.GradientBoostingClassifier(
    n_estimators=15,
    learning_rate=0.08,
    subsample=0.85,
    max_depth=3,
    random_state=42
)
gb_model.fit(X, y)
preds = gb_model.predict(X)
probs = gb_model.predict_proba(X)
print('GradientBoosting trained successfully!')
print(f'Sample predictions: {preds[:10]}')
```

## 4. Pure NumPy Model Evaluation & Metrics

```python
acc = ck.evaluation.accuracy_score(y, preds)
roc_auc = ck.evaluation.roc_auc_score(y, probs)
pr_auc = ck.evaluation.pr_auc_score(y, probs)
brier = ck.evaluation.brier_score_loss(y, probs)
cm, classes = ck.evaluation.confusion_matrix(y, preds)

print(f'Accuracy: {acc:.4f}')
print(f'ROC-AUC Score: {roc_auc:.4f}')
print(f'PR-AUC Score: {pr_auc:.4f}')
print(f'Brier Score Loss: {brier:.4f}')
print('Confusion Matrix:')
print(cm)
```

## 5. Permutation Feature Importance Explainability

```python
importance = ck.evaluation.permutation_importance(
    gb_model,
    X,
    y,
    scoring='accuracy',
    n_repeats=5,
    random_state=42
)
for col_idx, col_name in enumerate(['age', 'salary', 'credit_score']):
    print(f'{col_name:15s}: {importance["importances_mean"][col_idx]:.4f} +/- {importance["importances_std"][col_idx]:.4f}')
```

## 6. Zero Heavy Dependencies Verification
Audit running Python process to prove zero heavy libraries are imported.

```python
heavy_dependencies = ['scipy', 'sklearn', 'torch', 'tensorflow']
loaded = [lib for lib in heavy_dependencies if lib in sys.modules]
print(f'Loaded heavy dependencies: {loaded}')
assert len(loaded) == 0, f'Found heavy dependencies: {loaded}'
print('Verification SUCCESS: Chokkhu operates with 100% sovereign pure NumPy!')
```
