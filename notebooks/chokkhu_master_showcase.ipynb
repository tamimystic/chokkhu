# Chokkhu: The Sovereign Pure NumPy AI/ML Library
### Enterprise-Grade Machine Learning with Zero Heavy Dependencies
Chokkhu provides sovereign, minimal machine learning primitives (`pip install chokkhu`) with:
- **Zero Heavy Dependencies**: Absolutely no SciPy, Scikit-learn, PyTorch, or TensorFlow.
- **1-Line AutoML Pipeline**: Instant end-to-end data cleaning, preprocessing, model selection, and hyperparameter tuning.
- **Full Dynamic Parameter Support**: Complete flexibility for trees, boosting, neural nets, linear, and clustering models.

```python
import sys
import numpy as np
import pandas as pd
import chokkhu as ck
print(f'Chokkhu initialized: v{ck.__version__}')
```

## Section 1: End-to-End Tabular AutoML in 1 Line of Code

```python
np.random.seed(42)
df_raw = pd.DataFrame({
    'feature_x': np.random.randn(250),
    'feature_y': np.random.uniform(1.0, 100.0, 250),
    'category': np.random.choice(['Low', 'Medium', 'High'], 250),
    'label': np.random.choice([0, 1], 250, p=[0.5, 0.5])
})
df_raw.loc[::10, 'feature_x'] = np.nan

# 1-Line Sovereign AutoML
result = ck.auto_train(df_raw, target='label', task='classification', time_budget=10)
print(f'Optimal Model Selected: {result.best_model_name}')
print(f'Validation Score: {result.best_score:.4f}')
display(result.summary())
```

## Section 2: Model Families with Full Dynamic Hyperparameters

```python
X = df_raw.drop(columns=['label', 'category']).fillna(0).values
y = df_raw['label'].values

# Decision Tree with full dynamic parameter customization
dt = ck.models.ml.DecisionTreeClassifier(max_depth=4, min_samples_split=3, criterion='gini')
dt.fit(X, y)

# Random Forest with bootstrap & custom estimators
rf = ck.models.ml.RandomForestClassifier(n_estimators=10, max_depth=4, random_state=42)
rf.fit(X, y)

# Gradient Boosting with custom learning rate and subsampling
gb = ck.models.ml.GradientBoostingClassifier(n_estimators=10, learning_rate=0.05, subsample=0.8, random_state=42)
gb.fit(X, y)

# Multilayer Perceptron (Neural Network) in pure NumPy
mlp = ck.models.ml.MLPClassifier(hidden_layer_sizes=(32, 16), max_iter=50, learning_rate=0.01, random_state=42)
mlp.fit(X, y)

print('All model families trained with 100% dynamic hyperparameter control!')
```

## Section 3: Advanced Pure NumPy Evaluation & Explainability

```python
preds = gb.predict(X)
probs = gb.predict_proba(X)

print('Accuracy:', ck.evaluation.accuracy_score(y, preds))
print('ROC-AUC :', ck.evaluation.roc_auc_score(y, probs))
print('PR-AUC  :', ck.evaluation.pr_auc_score(y, probs))

# Permutation Explainability
imp = ck.evaluation.permutation_importance(gb, X, y, n_repeats=3, random_state=42)
print('Feature Importances (Mean):', imp['importances_mean'])
```
